```python
vector_0 = [5, 89, 79, 4, 56] 
vector_1 = [8, 9.4, 12, 78, 7.7]

scalaire = 0
for i in range(len(vector_0)):
    for a in range(len(vector_1)):
        if i == a:
            scalaire += vector_0[i] * vector_1[a]
print("Produit scalaire: ", scalaire)
```
